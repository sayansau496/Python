```python
import numpy as np
import pandas as pd

# Example DataFrame
df = pd.DataFrame(
    {
        "Name": ["XYZ", "LMN", "OPQ"],
        "Age":21,
        "Salary":25000,
        "Dept": ["HR", "Finance", "Sales"],
        "Bonus":4000,
    }
)





#print(df)
# New row data
new_row = ["ABC", 21, 21000, "IT", 2000]
df_new = pd.DataFrame(np.insert(df.values, 2, values=new_row, axis=0), columns=df.columns)
print(df)
print(df_new)
```

```python
# delete row using column-value filter
df = df.drop(df[df.Name == 'ABC'].index) # delete row, axis= 0
df
```

```python
# sort values - order values in dataframe by asc or desc order
df.sort_values('Name') #ascending order, by default
```

```python
# create a date column - date of joining
df['DOJ'] = ['2024-01-01', '2024-01-15', '2024-03-28']
df
```
